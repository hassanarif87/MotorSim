# PMSM simulation results

Plots from the C++ speed-control simulation. Regenerate the CSV with
`bazel run //:pmsm_sim` from `pmsmcpp`, then rerun this notebook.

```python
from pathlib import Path

import pandas as pd
import xy.pyplot as plt

csv_path = Path("pmsm_results.csv")
if not csv_path.is_file():
    csv_path = Path("..") / "pmsm_results.csv"
if not csv_path.is_file():
    csv_path = Path("pmsmcpp") / "pmsm_results.csv"

results = pd.read_csv(csv_path)
time_ms = results["time_s"].to_numpy() * 1000


def make_figure(rows, height):
    fig, axes = plt.subplots(rows, 1, sharex=True, figsize=(10, height))
    axes[-1].set_xlabel("Time (ms)")
    return fig, axes


def plot_signals(ax, columns, ylabel):
    for column, label, linestyle in columns:
        ax.plot(time_ms, results[column].to_numpy(), linestyle=linestyle, label=label)
    ax.set_ylabel(ylabel)
    ax.legend()
```

## Rotor motion

```python
fig, axes = make_figure(rows=2, height=4.8)
plot_signals(axes[0], [("omega_rad_s", "Speed", "-")], "Speed (rad/s)")
plot_signals(axes[1], [("theta_rad", "Position", "-")], "Position (rad)")
plt.close(fig)
fig
```

## Phase currents and dq tracking

```python
fig, axes = make_figure(rows=2, height=5.4)
plot_signals(
    axes[0],
    [("ia_a", "Phase A", "-"), ("ib_a", "Phase B", "-"), ("ic_a", "Phase C", "-")],
    "Phase current (A)",
)
plot_signals(
    axes[1],
    [
        ("id_a", "d measured", "-"),
        ("id_ref_a", "d reference", "--"),
        ("iq_a", "q measured", "-"),
        ("iq_ref_a", "q reference", "--"),
    ],
    "dq current (A)",
)
plt.close(fig)
fig
```

## Voltage commands and inverter output

```python
fig, axes = make_figure(rows=3, height=6.9)
plot_signals(
    axes[0],
    [("vd_v", "d command", "-"), ("vq_v", "q command", "-")],
    "dq voltage (V)",
)
plot_signals(
    axes[1],
    [("va_v", "Voltage A", "-"), ("vb_v", "Voltage B", "-"), ("vc_v", "Voltage C", "-")],
    "Phase voltage (V)",
)
plot_signals(
    axes[2],
    [("duty_a", "Duty A", "-"), ("duty_b", "Duty B", "-"), ("duty_c", "Duty C", "-")],
    "Duty cycle",
)
axes[2].set_ylim(-0.05, 1.05)
plt.close(fig)
fig
```
